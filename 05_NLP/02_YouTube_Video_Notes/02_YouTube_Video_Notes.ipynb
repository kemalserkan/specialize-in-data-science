### Not cikar


```python
import os
from dotenv import load_dotenv
load_dotenv()
from openai import OpenAI
from IPython.display import Markdown
client=OpenAI(base_url='https://openrouter.ai/api/v1',api_key=os.getenv('OPENROUTER_API_KEY'))
def notes_agent(t):
    r=client.chat.completions.create(model='openai/gpt-oss-120b:free',messages=[{'role':'user','content':'Turn this transcript into study notes.\n## Title ## Bullets ## Exam question\n'+t}])
    return Markdown(r.choices[0].message.content)
notes_agent(text)
```
