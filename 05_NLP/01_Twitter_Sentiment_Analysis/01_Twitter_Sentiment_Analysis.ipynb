# Tweet Sınıflandırma

Bu projede tweet'in nefret / saldırgan / normal olup olmadığını tahmin edeceğim.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/twitter.csv',encoding='latin-1')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

```python
df['class'].value_counts()
```

### Görselleştirme


```python
df['uzunluk']=df['tweet'].astype(str).str.len()
sns.boxplot(x='class',y='uzunluk',data=df)
plt.show()
```

### Boş veri


```python
df['tweet']=df['tweet'].fillna('')
```

### Feature Engineering


```python
from sklearn.feature_extraction.text import TfidfVectorizer
vec=TfidfVectorizer(max_features=2500,stop_words='english')
x=vec.fit_transform(df['tweet'].astype(str))
y=df['class']
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

### 3 Model


```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score,classification_report

for ad,m in [('NB',MultinomialNB()),('LogReg',LogisticRegression(max_iter=200)),('SVC',LinearSVC())]:
    m.fit(x_train,y_train)
    print(ad,accuracy_score(y_test,m.predict(x_test)))
```

```python
from sklearn.pipeline import make_pipeline
pipe=make_pipeline(TfidfVectorizer(max_features=2500,stop_words='english'),LogisticRegression(max_iter=200))
pipe.fit(df['tweet'].astype(str),y)
```

```python
import joblib
joblib.dump(pipe,'../../models/nlp_twitter_sentiment.joblib')
```
