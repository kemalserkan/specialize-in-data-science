# Müzik Türü Kümeleme

Bu projede Spotify şarkılarının ses özelliklerine bakarak tür grupları çıkaracağım.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/spotify_songs.csv')
df=df.sample(4000,random_state=42)  # set büyük, 4000 satır aldım
df.head()
```

### EDA


```python
df.info()
```

```python
df['playlist_genre'].value_counts()
```

### Görselleştirme


```python
sns.boxplot(x='playlist_genre',y='danceability',data=df)
plt.xticks(rotation=30)
plt.show()
```

### Boş veri


```python
num=df.select_dtypes('number').columns
df[num]=df[num].fillna(df[num].median())
```

### Feature Engineering


```python
feat=['danceability','energy','loudness','speechiness','acousticness','instrumentalness','liveness','valence','tempo']
from sklearn.preprocessing import StandardScaler
x=StandardScaler().fit_transform(df[feat])
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test=train_test_split(x,test_size=0.2,random_state=42)
```

### 3 Model


```python
from sklearn.cluster import KMeans,AgglomerativeClustering
from sklearn.mixture import GaussianMixture
from sklearn.metrics import silhouette_score

km=KMeans(n_clusters=6,random_state=42,n_init=10)
print('KMeans',round(silhouette_score(x,km.fit_predict(x)),3))
print('GMM',round(silhouette_score(x,GaussianMixture(6,random_state=42).fit_predict(x)),3))
print('Agglo',round(silhouette_score(x,AgglomerativeClustering(6).fit_predict(x)),3))
```

```python
import joblib
joblib.dump(km,'../../models/clustering_music_genres.joblib')
```

### Sonuç

Spotify özellikleriyle 6 küme denedim. Türler iç içe, silhouet düşük ama gruplar anlamlı. Hedefi kısmen tutturdum.
