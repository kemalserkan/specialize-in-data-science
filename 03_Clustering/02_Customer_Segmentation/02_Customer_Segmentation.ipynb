# Müşteri Segmentasyonu

Bu projede AVM müşterilerini yaş, gelir ve harcama skoruna göre gruplayacağım.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/customers.csv')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

```python
df.describe()
```

### Görselleştirme


```python
sns.scatterplot(x='Annual Income (k$)',y='Spending Score (1-100)',hue='Gender',data=df)
plt.show()
```

### Boş veri


```python
df=df.fillna(df.median(numeric_only=True))
```

### Feature Engineering


```python
df['Gender_num']=df['Gender'].map({'Male':1,'Female':0})
x=df[['Age','Annual Income (k$)','Spending Score (1-100)','Gender_num']]
from sklearn.preprocessing import StandardScaler
xs=StandardScaler().fit_transform(x)
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test=train_test_split(xs,test_size=0.2,random_state=42)
```

### 3 Model


```python
from sklearn.cluster import KMeans,AgglomerativeClustering,DBSCAN
from sklearn.metrics import silhouette_score

km=KMeans(n_clusters=5,random_state=42,n_init=10)
print('KMeans',silhouette_score(xs,km.fit_predict(xs)))
print('Agglo',silhouette_score(xs,AgglomerativeClustering(n_clusters=5).fit_predict(xs)))
lab=DBSCAN(eps=0.8,min_samples=5).fit_predict(xs)
print('DBSCAN kume sayisi',len(set(lab)))
```

```python
df['kume']=km.labels_
sns.scatterplot(x='Annual Income (k$)',y='Spending Score (1-100)',hue='kume',data=df,palette='tab10')
plt.show()
```

```python
import joblib
joblib.dump(km,'../../models/clustering_customer_segmentation.joblib')
```
