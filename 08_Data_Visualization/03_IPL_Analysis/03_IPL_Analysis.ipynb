# IPL 2022 Analizi

Mac tablosunu Day2 grafikleriyle okuyacagim. Tahmin modeli yok.


```python
import pandas as pd
pd.set_option('display.max_columns',100)
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### Data


```python
df=pd.read_csv('data/IPL2022.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

### Bos veri


```python
df['highscore']=df['highscore'].fillna(df['highscore'].median())
df['player_of_the_match']=df['player_of_the_match'].fillna('unknown')
```

### EDA


```python
df.describe()
```

```python
df['match_winner'].value_counts()
```

### matplotlib


```python
w=df['match_winner'].value_counts()
plt.bar(w.index,w.values)
plt.title('galibiyet')
plt.xticks(rotation=45)
plt.show()
```

```python
plt.pie(df['won_by'].value_counts(),labels=df['won_by'].value_counts().index,autopct='%1.0f%%')
plt.title('nasil kazandi')
plt.show()
```

```python
plt.hist(df['first_ings_score'],bins=12,alpha=0.6,label='1. inning')
plt.hist(df['second_ings_score'],bins=12,alpha=0.6,label='2. inning')
plt.legend()
plt.title('skor dagilimi')
plt.show()
```

```python
plt.scatter(df['first_ings_score'],df['second_ings_score'])
plt.xlabel('first')
plt.ylabel('second')
plt.title('iki inning')
plt.show()
```

```python
plt.boxplot(df['highscore'])
plt.title('en yuksek skor')
plt.show()
```

```python
plt.barh(df['top_scorer'].value_counts().head(8).index,df['top_scorer'].value_counts().head(8).values)
plt.title('top scorer')
plt.show()
```

```python
plt.plot(df['match_id'],df['first_ings_score'])
plt.title('mac sirasina gore 1. inning')
plt.show()
```

### seaborn


```python
sns.countplot(x=df['toss_decision'])
plt.show()
```

```python
sns.boxplot(x='stage',y='first_ings_score',data=df)
plt.xticks(rotation=30)
plt.show()
```

```python
sns.heatmap(df[['first_ings_score','first_ings_wkts','second_ings_score','second_ings_wkts','highscore']].corr(),annot=True)
plt.show()
```

### plotly


```python
px.bar(w.reset_index(name='n'),x='match_winner',y='n',title='kazanan')
```

```python
px.scatter(df,x='first_ings_score',y='second_ings_score',color='match_winner',title='skor')
```
