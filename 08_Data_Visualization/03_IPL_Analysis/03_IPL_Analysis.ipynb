# IPL 2022 Analizi

Bu projede IPL maçlarını görselleştireceğim. Ayrıca kazananı 3 modelle tahmin etmeyi denedim.


```python
import pandas as pd
pd.set_option('display.max_columns',100)
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/IPL2022.csv')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

### Görselleştirme


```python
df['match_winner'].value_counts().plot(kind='bar')
plt.title('galibiyet')
plt.show()
```

```python
sns.scatterplot(x='first_ings_score',y='second_ings_score',hue='match_winner',data=df,legend=False)
plt.show()
```

### Boş veri


```python
df['highscore']=df['highscore'].fillna(df['highscore'].median())
df['player_of_the_match']=df['player_of_the_match'].fillna('unknown')
```

### Feature Engineering


```python
df['hedef']=(df['match_winner']==df['team1']).astype(int)
x=df[['first_ings_score','first_ings_wkts','second_ings_score','second_ings_wkts','highscore']].copy()
x=pd.get_dummies(x.join(df[['toss_decision']]),drop_first=True)
y=df['hedef']
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

### 3 Model


```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score

for ad,m in [('LogReg',LogisticRegression(max_iter=200)),('DT',DecisionTreeClassifier(random_state=42)),('RF',RandomForestClassifier(random_state=42))]:
    m.fit(x_train,y_train)
    print(ad,accuracy_score(y_test,m.predict(x_test)))
```

### Sonuç

GT ve RR öne çıkmış. Skorlar maçı zaten anlatıyor, model yüksek çıkar. EDA hedefini tutturdum.
