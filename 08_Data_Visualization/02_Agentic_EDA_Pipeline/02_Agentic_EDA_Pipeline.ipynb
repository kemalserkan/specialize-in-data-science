# Ulke Karsilastirmasi

Corona tablosundan Turkiye ve bir kac ulkeyi cizdim.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### Data Dosyasını Okuma


```python
df=pd.read_csv('data/corona_dat.csv')
df.head()
```

```python
df.shape
```

```python
df.isnull().sum().sum()
```

### Kolonlar


```python
df['date']=pd.to_datetime(df['date'])
tab=df.set_index('date')[['Turkey','Italy','Germany','Spain','United Kingdom']]
```

### Veri Görselleştirme


```python
plt.plot(tab.index,tab['Turkey'])
plt.title('Turkey')
plt.show()
```

```python
tab.plot()
plt.title('secilen ulkeler')
plt.show()
```

```python
sns.heatmap(tab.corr(),annot=True)
plt.show()
```

```python
px.line(tab.reset_index(),x='date',y=['Turkey','Italy','Germany','Spain','United Kingdom'])
```

### Yorum


```python
import os
from dotenv import load_dotenv
load_dotenv()
from openai import OpenAI
from IPython.display import Markdown
client=OpenAI(base_url='https://openrouter.ai/api/v1',api_key=os.getenv('OPENROUTER_API_KEY'))
def eda_agent(stats):
    r=client.chat.completions.create(model='openai/gpt-oss-120b:free',messages=[{'role':'user','content':'EDA agent. Format: ## Trend ## Risk ## 3 bullets\n'+stats}])
    return Markdown(r.choices[0].message.content)
eda_agent(tab.describe().round(1).to_string())
```
