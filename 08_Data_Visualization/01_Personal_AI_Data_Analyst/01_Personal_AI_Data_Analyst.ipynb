# Market Satis Analizi

Supermarket csv. Once grafikle bakayim, sonra kisa bir ozet alirim.


```python
import pandas as pd
pd.set_option('display.max_columns',100)
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### Data Dosyasını Okuma


```python
df=pd.read_csv('data/supermarket.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

### Boş Verileri Doldurma


```python
df['Rating']=df['Rating'].fillna(df['Rating'].median())
```

### EDA


```python
df.describe()
```

### matplotlib


```python
plt.hist(df['Total'],bins=30)
plt.title('satis tutari')
plt.show()
```

```python
plt.bar(df['City'].value_counts().index,df['City'].value_counts().values)
plt.title('sehir')
plt.show()
```

```python
plt.pie(df['Payment'].value_counts(),labels=df['Payment'].value_counts().index,autopct='%1.0f%%')
plt.title('odeme')
plt.show()
```

```python
plt.scatter(df['Unit price'],df['Rating'],alpha=0.4)
plt.xlabel('fiyat')
plt.ylabel('rating')
plt.show()
```

```python
plt.boxplot(df['Quantity'])
plt.title('adet')
plt.show()
```

### seaborn


```python
sns.countplot(x=df['Product line'],hue=df['Gender'])
plt.xticks(rotation=45)
plt.show()
```

```python
sns.boxplot(x='Branch',y='Rating',data=df)
plt.show()
```

```python
sns.heatmap(df[['Unit price','Quantity','Total','Rating']].corr(),annot=True)
plt.show()
```

### plotly


```python
px.bar(df.groupby('Product line')['Total'].sum().reset_index(),x='Product line',y='Total')
```

```python
px.scatter(df,x='Total',y='Rating',color='City')
```
