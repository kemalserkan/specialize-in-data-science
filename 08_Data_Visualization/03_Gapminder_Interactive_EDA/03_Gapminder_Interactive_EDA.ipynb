# Gapminder

Plotly'nin kendi seti. Hover ile ulke bakmak kolay.


```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
```

### Data


```python
df=px.data.gapminder()
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

### matplotlib


```python
y=df[df['year']==2007]
plt.scatter(y['gdpPercap'],y['lifeExp'])
plt.xlabel('gdp')
plt.ylabel('lifeExp')
plt.show()
```

```python
plt.hist(y['lifeExp'],bins=20)
plt.show()
```

```python
plt.barh(y.groupby('continent')['pop'].sum().index,y.groupby('continent')['pop'].sum().values)
plt.show()
```

### seaborn


```python
sns.boxplot(x='continent',y='lifeExp',data=y)
plt.show()
```

```python
sns.pairplot(y[['gdpPercap','lifeExp','pop']],diag_kind='hist')
```

### plotly


```python
#kopyala yapıştır hocadan
fig=px.scatter(df.query('year==2007'),x='gdpPercap',y='lifeExp',
         size='pop',color='continent',
                 hover_name='country',log_x=True,size_max=60)
fig.show()
```

```python
px.sunburst(y,path=['continent','country'],values='pop')
```

```python
tips=px.data.tips()
px.scatter(tips,x='total_bill',y='tip',color='sex')
```
