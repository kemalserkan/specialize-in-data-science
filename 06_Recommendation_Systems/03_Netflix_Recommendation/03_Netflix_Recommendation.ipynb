# Netflix İçerik Önerisi

Bu projede Netflix'te olan filmlere benzer içerik önereceğim.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/MoviesOnStreamingPlatforms_updated.csv')
df=df[df['Netflix']==1].copy()
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

### Görselleştirme


```python
df['Genres'].fillna('').str.split(',').explode().value_counts().head(10).plot(kind='bar')
plt.show()
```

### Boş veri


```python
df['Genres']=df['Genres'].fillna('')
df['IMDb']=df['IMDb'].fillna(df['IMDb'].median())
df['Age']=df['Age'].fillna('Unknown')
```

### Feature Engineering


```python
from sklearn.feature_extraction.text import TfidfVectorizer
df['soup']=df['Genres']+' '+df['Age']
vec=TfidfVectorizer()
mat=vec.fit_transform(df['soup'])
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
_,_=train_test_split(df.index,test_size=0.2,random_state=42)
```

### 3 yaklaşım


```python
from sklearn.metrics.pairwise import cosine_similarity,linear_kernel,pairwise_distances
import numpy as np
sim=cosine_similarity(mat)
i=0
print(df.iloc[np.argsort(sim[i])[::-1][1:6]]['Title'].tolist())
```

```python
import joblib
joblib.dump({'vec':vec,'titles':df['Title'].tolist()},'../../models/recsys_netflix.joblib')
```
