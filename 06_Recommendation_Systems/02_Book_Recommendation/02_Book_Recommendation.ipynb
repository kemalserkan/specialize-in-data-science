# Kitap Öneri Sistemi

Bu projede bir kitaba benzer kitapları önereceğim.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/books.csv')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

### Görselleştirme


```python
sns.histplot(df['average_rating'],bins=20)
plt.show()
```

### Boş veri


```python
df['authors']=df['authors'].fillna('unknown')
df['original_title']=df['original_title'].fillna(df['title'])
df['average_rating']=df['average_rating'].fillna(df['average_rating'].median())
df['original_publication_year']=df['original_publication_year'].fillna(df['original_publication_year'].median())
```

### Feature Engineering


```python
from sklearn.feature_extraction.text import TfidfVectorizer
df['soup']=df['authors'].astype(str)+' '+df['original_title'].astype(str)
vec=TfidfVectorizer(max_features=1500)
mat=vec.fit_transform(df['soup'])
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
_,_=train_test_split(df.index,test_size=0.2,random_state=42)
```

### 3 yaklaşım


```python
from sklearn.metrics.pairwise import cosine_similarity,linear_kernel
from sklearn.neighbors import NearestNeighbors
import numpy as np
M=mat[:2000]
print('cosine ornek',df.iloc[np.argsort(cosine_similarity(M)[0])[::-1][1:6]]['title'].tolist())
nn=NearestNeighbors(metric='cosine',n_neighbors=6)
nn.fit(M)
print('knn',nn.kneighbors(M[0],return_distance=False))
```

```python
import joblib
joblib.dump({'vec':vec},'../../models/recsys_book.joblib')
```
