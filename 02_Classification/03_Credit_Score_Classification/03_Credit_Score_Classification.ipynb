```python
import pandas as pd
pd.set_option('display.max_columns',100)
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/cs-training.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

### Bos veri


```python
df['MonthlyIncome']=df['MonthlyIncome'].fillna(df['MonthlyIncome'].median())
df['NumberOfDependents']=df['NumberOfDependents'].fillna(0)
```

### EDA


```python
df['SeriousDlqin2yrs'].value_counts()
```

```python
sns.countplot(x=df['SeriousDlqin2yrs'])
plt.show()
```

```python
sns.heatmap(df.corr(numeric_only=True),cmap='coolwarm')
plt.show()
```

### Feature


```python
y=df['SeriousDlqin2yrs']
x=df.drop(columns=['SeriousDlqin2yrs'])
```

### Train test


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

### Modeller


```python
from sklearn.linear_model import LogisticRegression
from sklearn.tree import DecisionTreeClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score
```

```python
m1=LogisticRegression(max_iter=300)
m1.fit(x_train,y_train)
print('LogReg',accuracy_score(y_test,m1.predict(x_test)))
```

```python
m2=DecisionTreeClassifier(random_state=42,max_depth=8)
m2.fit(x_train,y_train)
print('DT',accuracy_score(y_test,m2.predict(x_test)))
```

```python
m3=RandomForestClassifier(random_state=42)
m3.fit(x_train,y_train)
print('RF',accuracy_score(y_test,m3.predict(x_test)))
```
