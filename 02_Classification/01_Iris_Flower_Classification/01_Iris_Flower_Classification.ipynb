# Iris Çiçek Sınıflandırma

Bu projede çiçeğin yaprak ölçüleriyle türünü tahmin edeceğim. Klasik bir set, 3 sınıf var.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/IRIS.csv')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

```python
df['species'].value_counts()
```

### Görselleştirme


```python
sns.pairplot(df,hue='species')
plt.show()
```

### Boş veri


```python
df=df.fillna(df.median(numeric_only=True))
```

### Feature Engineering


```python
x=df.drop('species',axis=1)
y=df['species']
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

### 3 Model


```python
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report
```

```python
m1=LogisticRegression(max_iter=200)
m1.fit(x_train,y_train)
print('LogReg',accuracy_score(y_test,m1.predict(x_test)))
```

```python
m2=KNeighborsClassifier()
m2.fit(x_train,y_train)
print('KNN',accuracy_score(y_test,m2.predict(x_test)))
```

```python
m3=RandomForestClassifier(random_state=42)
m3.fit(x_train,y_train)
print('RF',accuracy_score(y_test,m3.predict(x_test)))
```

```python
rf=RandomForestClassifier(random_state=42)
rf.fit(x_train,y_train)
print(classification_report(y_test,rf.predict(x_test)))
```

```python
import joblib
joblib.dump(rf,'../../models/classification_iris.joblib')
```
