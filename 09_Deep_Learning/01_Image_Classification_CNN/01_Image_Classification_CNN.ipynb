# Görüntü Sınıflandırma (NN)

Bu projede el yazısı rakamları sınıflandıracağım. Ortamda TF takıldığı için sklearn MLP kullandım, 3 model karşılaştırdım.


```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.datasets import load_digits
```

### Data


```python
d=load_digits()
X=d.data; y=d.target
print(X.shape)
```

### EDA


```python
print(pd.DataFrame(X).isnull().sum().sum())
print(pd.Series(y).value_counts().sort_index())
```

### Görselleştirme


```python
plt.imshow(d.images[3],cmap='gray')
plt.axis('off')
plt.show()
```

### Boş veri


```python
import numpy as np
X=np.nan_to_num(X)
```

### Feature Engineering


```python
from sklearn.preprocessing import MinMaxScaler
Xs=MinMaxScaler().fit_transform(X)
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(Xs,y,test_size=0.2,random_state=42,stratify=y)
```

### 3 Model


```python
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report

for ad,m in [('MLP',MLPClassifier(hidden_layer_sizes=(64,32),max_iter=300,random_state=42)),('LogReg',LogisticRegression(max_iter=200)),('RF',RandomForestClassifier(random_state=42))]:
    m.fit(x_train,y_train)
    print(ad,accuracy_score(y_test,m.predict(x_test)))
```

```python
mlp=MLPClassifier(hidden_layer_sizes=(64,32),max_iter=300,random_state=42)
mlp.fit(x_train,y_train)
print(classification_report(y_test,mlp.predict(x_test)))
```

```python
import joblib
joblib.dump(mlp,'../../models/dl_image_cnn.joblib')
```
