# Computer Vision with Deep Learning

Day5'teki MNIST CNN. sklearn MLP yok.


```python
import tensorflow as tf
mnist=tf.keras.datasets.mnist
(train_images,train_labels),(test_images,test_labels)=mnist.load_data()
train_images.shape
```

```python
import matplotlib.pyplot as plt
plt.imshow(train_images[0],cmap='gray')
```

```python
plt.imshow(train_images[10000],cmap='gray')
```

### Normalizasyon


```python
train_images=train_images/255.0
test_images=test_images/255.0
```

### CNN - ders kodu


```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Conv2D, InputLayer, Reshape, MaxPooling2D, Flatten, Dropout, BatchNormalization
```

```python
model=Sequential()
model.add(InputLayer(input_shape=(28,28)))
model.add(Reshape(target_shape=(28,28,1)))
model.add(Conv2D(filters=12,kernel_size=(3,3),activation='relu'))
model.add(MaxPooling2D(pool_size=(2,2)))
model.add(Flatten())
model.add(Dense(10))
model.compile(optimizer='adam',loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),metrics=['accuracy'])
```

```python
model.summary()
```

```python
model.fit(train_images,train_labels,validation_split=.10,epochs=5)
```

```python
loss,accuracy=model.evaluate(test_images,test_labels)
accuracy
```

### 2. model CNN + Dropout


```python
model2=Sequential()
model2.add(InputLayer(input_shape=(28,28)))
model2.add(Reshape(target_shape=(28,28,1)))
model2.add(Conv2D(filters=16,kernel_size=(3,3),activation='relu'))
model2.add(MaxPooling2D(pool_size=(2,2)))
model2.add(Dropout(0.2))
model2.add(Flatten())
model2.add(Dense(10))
model2.compile(optimizer='adam',loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),metrics=['accuracy'])
model2.fit(train_images,train_labels,validation_split=.10,epochs=4,verbose=1)
print(model2.evaluate(test_images,test_labels))
```

### 3. model Dense (Day5 pima tarzi, duz resim)


```python
xtr=train_images.reshape(-1,28*28)
xte=test_images.reshape(-1,28*28)
model3=Sequential()
model3.add(Dense(80,activation='relu'))
model3.add(Dense(64,activation='relu'))
model3.add(Dense(10))
model3.compile(optimizer='adam',loss=tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),metrics=['accuracy'])
model3.fit(xtr,train_labels,validation_split=.10,epochs=4,verbose=1)
print(model3.evaluate(xte,test_labels))
```
