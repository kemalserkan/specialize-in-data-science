# Metin Sınıflandırma (NN)

Bu projede Flipkart yorumunun puanını (yüksek/düşük) tahmin edeceğim.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data


```python
df=pd.read_csv('data/flipkart_reviews.csv')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

```python
df['Rating'].value_counts()
```

### Görselleştirme


```python
df['Rating'].value_counts().sort_index().plot(kind='bar')
plt.show()
```

### Boş veri


```python
df['Review']=df['Review'].fillna('')
df=df.dropna(subset=['Rating'])
```

### Feature Engineering


```python
df['yildiz']=(df['Rating']>=4).astype(int)
from sklearn.feature_extraction.text import TfidfVectorizer
vec=TfidfVectorizer(max_features=2500,stop_words='english')
x=vec.fit_transform(df['Review'].astype(str))
y=df['yildiz']
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42,stratify=y)
```

### 3 Model


```python
from sklearn.neural_network import MLPClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import MultinomialNB
from sklearn.metrics import accuracy_score,classification_report

for ad,m in [('MLP',MLPClassifier(hidden_layer_sizes=(64,),max_iter=200,random_state=42)),('LogReg',LogisticRegression(max_iter=200)),('NB',MultinomialNB())]:
    m.fit(x_train,y_train)
    print(ad,accuracy_score(y_test,m.predict(x_test)))
```

```python
from sklearn.pipeline import make_pipeline
pipe=make_pipeline(TfidfVectorizer(max_features=2500,stop_words='english'),LogisticRegression(max_iter=200))
pipe.fit(df['Review'].astype(str),y)
```

```python
import joblib
joblib.dump(pipe,'../../models/dl_text_nn.joblib')
```

### Sonuç

Flipkart yorumlarında 5 yıldız çok. LogReg yeterli, MLP biraz ağır. Hedefi tutturdum.
