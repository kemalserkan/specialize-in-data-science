# NLP with Deep Learning

Flipkart yorumu. Day7'deki gibi CountVectorizer + Keras Sequential Dense.


```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout
from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import CountVectorizer
```

### Data


```python
df=pd.read_csv('data/flipkart_reviews.csv')
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

### Bos veri


```python
df['Review']=df['Review'].fillna('')
df=df.dropna(subset=['Rating'])
```

### EDA


```python
sns.countplot(x=df['Rating'])
plt.show()
```

### Feature


```python
y=(df['Rating']>=4).astype(int)
cv=CountVectorizer(max_features=2000,stop_words='english')
x=cv.fit_transform(df['Review'].astype(str)).toarray()
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
x_train.shape
```

### 1. model Day7


```python
model=Sequential([
    Dense(256,activation='relu'),
    Dense(128,activation='relu'),
    Dense(64,activation='relu'),
    Dense(32,activation='relu'),
    Dense(1,activation='sigmoid') # binary 0-1 ise sigmoid
])
model.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
```

```python
model.fit(x_train,y_train,batch_size=32,validation_data=(x_test,y_test),verbose=2,epochs=6)
```

```python
model.evaluate(x_test,y_test)
```

### 2. model daha dar


```python
m2=Sequential()
m2.add(Dense(80,activation='relu'))
m2.add(Dense(30,activation='relu'))
m2.add(Dense(1,activation='sigmoid'))
m2.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
m2.fit(x_train,y_train,validation_split=0.1,epochs=5,verbose=1)
print(m2.evaluate(x_test,y_test))
```

### 3. model Dropout


```python
m3=Sequential()
m3.add(Dense(128,activation='relu'))
m3.add(Dropout(0.3))
m3.add(Dense(1,activation='sigmoid'))
m3.compile(loss='binary_crossentropy',optimizer='adam',metrics=['accuracy'])
m3.fit(x_train,y_train,validation_split=0.1,epochs=5,verbose=1)
print(m3.evaluate(x_test,y_test))
```
