# El Yazısı / Jest Benzeri Görüntü Sınıflandırma

Bu projede el ile üretilmiş rakam görüntülerini (sklearn digits) sınıflandıracağım. Jest tanıma ile aynı mantık: görüntü -> özellik -> sınıf.


```python
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
from sklearn.datasets import load_digits
```

### Data


```python
d=load_digits()
X=d.data; y=d.target
print(X.shape)
```

### EDA


```python
print('bos',pd.DataFrame(X).isnull().sum().sum())
print(pd.Series(y).value_counts().sort_index())
```

### Görselleştirme


```python
plt.imshow(d.images[0],cmap='gray')
plt.title(str(y[0]))
plt.axis('off')
plt.show()
```

### Boş veri


```python
import numpy as np
X=np.nan_to_num(X)
```

### Feature Engineering


```python
from sklearn.preprocessing import MinMaxScaler
Xs=MinMaxScaler().fit_transform(X)
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(Xs,y,test_size=0.2,random_state=42,stratify=y)
```

### 3 Model


```python
from sklearn.neighbors import KNeighborsClassifier
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import accuracy_score,classification_report
```

```python
from sklearn.linear_model import LogisticRegression
m1=LogisticRegression(max_iter=200)
m1.fit(x_train,y_train)
print('LogReg',accuracy_score(y_test,m1.predict(x_test)))
```

```python
m2=KNeighborsClassifier()
m2.fit(x_train,y_train)
print('KNN',accuracy_score(y_test,m2.predict(x_test)))
```

```python
m3=RandomForestClassifier(random_state=42)
m3.fit(x_train,y_train)
print('RF',accuracy_score(y_test,m3.predict(x_test)))
```

```python
knn=KNeighborsClassifier()
knn.fit(x_train,y_train)
print(classification_report(y_test,knn.predict(x_test)))
```

```python
import joblib
joblib.dump(knn,'../../models/cv_hand_gesture.joblib')
```
