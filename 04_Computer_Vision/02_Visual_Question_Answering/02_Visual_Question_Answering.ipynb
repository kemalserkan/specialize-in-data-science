```python
import matplotlib.pyplot as plt
import cv2
```

### Fotolar


```python
img=cv2.imread('data/typewritten.png')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
plt.imshow(cv2.cvtColor(cv2.imread('data/Sunflowers.jpg'),cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

### Yaziyi oku


```python
import ollama
res=ollama.chat(model='llama3.2-vision',messages=[{
    'role':'user',
    'content':'extract the exact text from this image. Nothing less, nothing more',
    'images':['data/typewritten.png']
}])
print(res['message']['content'])
```

```python
res=ollama.chat(model='llama3.2-vision',messages=[{
    'role':'user',
    'content':'What is in this photo? One short sentence.',
    'images':['data/Sunflowers.jpg']
}])
print(res['message']['content'])
```
