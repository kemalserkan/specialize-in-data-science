# Görüntüde Nesne Sayma

Bu projede OpenCV ile bir fotoğraftaki nesneleri sayacağım. Hedefim sayıyı doğru tutturmak.


```python
import cv2
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import warnings
warnings.filterwarnings('ignore')
```

### Data


```python
img=cv2.imread('data/smarties.png')
img=cv2.cvtColor(img,cv2.COLOR_BGR2RGB)
plt.imshow(img)
plt.axis('off')
plt.show()
```

### EDA


```python
print(img.shape,img.dtype)
print('bos piksel yok',np.isnan(img).sum())
```

### Görselleştirme


```python
gray=cv2.cvtColor(img,cv2.COLOR_RGB2GRAY)
plt.imshow(gray,cmap='gray')
plt.show()
```

### Boş / ön işleme


```python
blur=cv2.medianBlur(gray,5)
```

### Feature Engineering

3 yöntem: threshold, canny, blob.


```python
_,th=cv2.threshold(blur,0,255,cv2.THRESH_BINARY+cv2.THRESH_OTSU)
canny=cv2.Canny(blur,50,150)
params=cv2.SimpleBlobDetector_Params()
params.filterByArea=True
params.minArea=50
blob=cv2.SimpleBlobDetector_create(params)
```

### 3 Model / yöntem


```python
c1,_=cv2.findContours(th,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
c2,_=cv2.findContours(canny,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
kp=blob.detect(blur)
print('threshold',len(c1),'canny',len(c2),'blob',len(kp))
```

```python
out=img.copy()
cv2.drawContours(out,c1,-1,(255,0,0),2)
plt.imshow(out)
plt.title('threshold sayim')
plt.axis('off')
plt.show()
```

```python
import joblib
joblib.dump({'threshold':len(c1),'canny':len(c2),'blob':len(kp)},'../../models/cv_count_objects.joblib')
```
