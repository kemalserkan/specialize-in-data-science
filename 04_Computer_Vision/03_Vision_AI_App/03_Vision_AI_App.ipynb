# OpenCV Deneme

Resmi acip griye cevirdim, kenarlari Canny ile cikardim.


```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

### Data


```python
img=cv2.imread('data/Sunflowers.jpg')
plt.imshow(cv2.cvtColor(img,cv2.COLOR_BGR2RGB))
plt.title('orijinal')
plt.axis('off')
plt.show()
```

# Otonom araçlar yollarını nasıl bulur: Canny - Border sınır


```python
gray=cv2.cvtColor(img,cv2.COLOR_BGR2GRAY) # Siyah beyaza çeviren kod
plt.imshow(gray,cmap='gray')
plt.axis('off')
plt.show()
```

```python
blur=cv2.GaussianBlur(gray,(5,5),0)
edge=cv2.Canny(blur,80,160)
plt.imshow(edge,cmap='gray')
plt.title('Canny')
plt.axis('off')
plt.show()
```

### Kontur say


```python
cnts,hir=cv2.findContours(edge,cv2.RETR_EXTERNAL,cv2.CHAIN_APPROX_SIMPLE)
ciz=img.copy()
cv2.drawContours(ciz,cnts,-1,(0,255,0),2)
plt.imshow(cv2.cvtColor(ciz,cv2.COLOR_BGR2RGB))
plt.title(f'kontur {len(cnts)}')
plt.axis('off')
plt.show()
```
