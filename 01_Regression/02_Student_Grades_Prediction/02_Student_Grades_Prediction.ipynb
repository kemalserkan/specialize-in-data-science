# Öğrenci Not Tahmini

Bu projede öğrencinin ders sayısı ve çalışma süresine bakarak notunu tahmin edeceğim. Amacım notu makul bir hata ile tutturmak.


```python
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
import seaborn as sns
```

### Data Okuma


```python
df=pd.read_csv('data/Student_Marks.csv')
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

```python
df.describe()
```

### Veri Görselleştirme


```python
sns.pairplot(df)
plt.show()
```

```python
sns.heatmap(df.corr(),annot=True)
plt.show()
```

### Boş Veri


```python
df=df.fillna(df.median())
```

### Feature Engineering


```python
df['saat_ders']=df['time_study']/(df['number_courses']+1)
x=df.drop('Marks',axis=1)
y=df['Marks']
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

### 3 Model


```python
from sklearn.linear_model import LinearRegression,Ridge
from sklearn.ensemble import RandomForestRegressor
from sklearn.metrics import r2_score,mean_absolute_error
```

```python
m1=LinearRegression()
m1.fit(x_train,y_train)
print('LR',round(r2_score(y_test,m1.predict(x_test)),3))
```

```python
m2=Ridge()
m2.fit(x_train,y_train)
print('Ridge',round(r2_score(y_test,m2.predict(x_test)),3))
```

```python
m3=RandomForestRegressor(random_state=42)
m3.fit(x_train,y_train)
print('RF',round(r2_score(y_test,m3.predict(x_test)),3))
```

### Feature Importance ve Residual


```python
rf=RandomForestRegressor(random_state=42)
rf.fit(x_train,y_train)
pred=rf.predict(x_test)

print(pd.Series(rf.feature_importances_,index=x.columns).sort_values(ascending=False))

plt.scatter(pred,y_test-pred)
plt.axhline(0,color='r')
plt.show()
```

```python
import joblib
joblib.dump(rf,'../../models/regression_student_grades.joblib')
```

### Sonuç

Student_Marks verisinde çalışma süresi nota çok bağlı. Linear / Ridge burada yeterli geldi, RF de yakın. Hedefi tutturdum, küçük sette R2 yüksek çıktı.
