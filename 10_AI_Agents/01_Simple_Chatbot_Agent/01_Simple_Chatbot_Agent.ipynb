# Basit Chatbot Agent

Bu projede intent dosyasından niyet tahmini yapan bir chatbot kuracağım. 3 sınıflandırıcı denedim.


```python
import json
import pandas as pd
import warnings
warnings.filterwarnings('ignore')
import matplotlib.pyplot as plt
```

### Data


```python
intents=json.load(open('data/intents.json'))
rows=[]
for it in intents['intents']:
    for p in it['patterns']:
        rows.append({'text':p,'tag':it['tag']})
df=pd.DataFrame(rows)
df.head()
```

### EDA


```python
df.info()
df.isnull().sum()
```

```python
df['tag'].value_counts()
```

### Görselleştirme


```python
df['tag'].value_counts().plot(kind='bar')
plt.show()
```

### Boş veri


```python
df['text']=df['text'].fillna('')
```

### Feature Engineering


```python
from sklearn.feature_extraction.text import TfidfVectorizer
vec=TfidfVectorizer()
x=vec.fit_transform(df['text'])
y=df['tag']
```

### Train Test Split


```python
from sklearn.model_selection import train_test_split
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.25,random_state=42)
```

### 3 Model


```python
from sklearn.naive_bayes import MultinomialNB
from sklearn.linear_model import LogisticRegression
from sklearn.svm import LinearSVC
from sklearn.metrics import accuracy_score

for ad,m in [('NB',MultinomialNB()),('LogReg',LogisticRegression(max_iter=200)),('SVC',LinearSVC())]:
    m.fit(x_train,y_train)
    print(ad,accuracy_score(y_test,m.predict(x_test)))
```

```python
from sklearn.pipeline import make_pipeline
pipe=make_pipeline(TfidfVectorizer(),LinearSVC())
pipe.fit(df['text'],y)
print(pipe.predict(['hello there'])[0])
```

```python
import joblib
joblib.dump({'pipe':pipe,'intents':intents},'../../models/agent_chatbot.joblib')
```
