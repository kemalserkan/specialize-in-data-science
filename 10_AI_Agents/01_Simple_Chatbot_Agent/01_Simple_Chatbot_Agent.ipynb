```python
import os
from dotenv import load_dotenv
load_dotenv()
from openai import OpenAI
from IPython.display import Markdown
```

### Agent (OpenRouter - dersteki client)


```python
client=OpenAI(
  base_url='https://openrouter.ai/api/v1',
  api_key=os.getenv('OPENROUTER_API_KEY'),
)
```

```python
def lab_agent(query):
    prompt=f'''
You are an autonomous AI lab assistant.
Help design experiments safely and logically.
Format: ## Hypothesis, Experiment, Data Needed, Failure Modes.

Task:
{query}
'''
    r=client.chat.completions.create(
        model='openai/gpt-oss-120b:free',
        messages=[{'role':'user','content':prompt}]
    )
    return Markdown(r.choices[0].message.content)
```

```python
lab_agent('Design an experiment to test a new battery material')
```

### Mars planner agent


```python
def mars_agent(query):
    prompt=f'''
You are a Mars colony systems planner.
Design compact survival strategies.
Format: ## Mission Plan, Oxygen, Water, Energy, Risks.

Request:
{query}
'''
    r=client.chat.completions.create(
        model='openai/gpt-oss-120b:free',
        messages=[{'role':'user','content':prompt}]
    )
    return Markdown(r.choices[0].message.content)

mars_agent('Plan a 30-person Mars base for the first 180 days')
```
